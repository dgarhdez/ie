## Exercise 1

Create a list with all the students that have a second last name.

```python
has_second_name = []

for student_dict in students_list:
    full_name = student_dict["name"]
    names_list = full_name.split()
    len_names_list = len(names_list)
    if len_names_list > 2:
        has_second_name.append(full_name)

has_second_name
```

## Exercise 2
 
Create a list with all the students whose last name is fully included in their email.

```python
has_second_name = []

for student_dict in students_list:
    full_name = student_dict["name"]
    names_list = full_name.split()
    len_names_list = len(names_list)
    if len_names_list >= 2:
        if "".join(names_list[1:]).lower() in student_dict["email"]:
            has_second_name.append(full_name)

has_second_name
```

## Exercise 5

Create a dictionary with the number of students coming from each background

```python
[student["background"] for student in students_list]
```

```python
unique_background = set([student["background"] for student in students_list])

bg_dict = dict()

for bg in unique_background:
    counter = 0
    for student in students_list:
        if student["background"] == bg:
            counter += 1
    bg_dict[bg] = counter

bg_dict
```

```python
bg_dict = dict()  # {bg: counter}

for student in students_list:
    background = student["background"]
    if (
        background in bg_dict.keys()
    ):  # if the background already exists as a key in bg_dict
        bg_dict[background] = bg_dict[background] + 1  # increase 1 the counter
    else:  # if the key doesnt exist yet
        bg_dict[background] = 1  # create the key and assign 1 to it

bg_dict
```

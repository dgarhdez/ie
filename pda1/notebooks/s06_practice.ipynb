### Exercise 6

Given 3 lists, create a dictionary where the first list are the keys, and and second and third list should be broken down as tuples associated to each key:

```python
list_keys = ['a', 'b', 'c']
list_values_1 = [1, 2, 3]
list_values_2 = [10, 20, 30]
```

For example, the first pair of the dictionary should be `'a': (1, 10)`

````python
list_keys = ['a', 'b', 'c']
list_values_1 = [1, 2, 3]
list_values_2 = [10, 20, 30]

final_dict = dict(zip(list_keys, zip(list_values_1, list_values_2)))

final_dict
````

### Exercise 13

Given the abc, create a dictionary where keys are the letters in odd positions and values are the letters in even positions. For example:

```python
{
    'a': 'b',
    'c': 'd',
    'e': 'f',
    ...
}
```

````python
abc = 'abcdefghijklmnopqrstuvwxyz'

# odds positions
odds = abc[0::2]

# even
evens = abc[1::2]

dict(zip(odds, evens))
````

# Practice: Spotify

This dataset is a list of dictionaries, where each dictionary represents a song played on Spotify. Each dictionary contains the following keys:

- `ts`: Timestamp when the song was played (ISO 8601 format)
- `username`: Spotify username of the listener
- `platform`: Operating system and device information where the song was played
- `ms_played`: Duration the song was played in milliseconds
- `conn_country`: Country code where the connection originated (2-letter ISO code)
- `ip_addr_decrypted`: Decrypted IP address of the listener
- `user_agent_decrypted`: Decrypted user agent string (browser/app information)
- `master_metadata_track_name`: Name of the track/song
- `master_metadata_album_artist_name`: Name of the artist
- `master_metadata_album_album_name`: Name of the album
- `spotify_track_uri`: Unique Spotify URI identifier for the track
- `episode_name`: Name of the podcast episode (if applicable, otherwise None)
- `episode_show_name`: Name of the podcast show (if applicable, otherwise None)
- `spotify_episode_uri`: Unique Spotify URI for podcast episodes (if applicable, otherwise None)
- `reason_start`: Reason why the track started playing
- `reason_end`: Reason why the track stopped playing
- `shuffle`: Boolean indicating if shuffle mode was enabled
- `skipped`: Information about whether the track was skipped (None if not skipped)
- `offline`: Boolean indicating if the track was played offline
- `offline_timestamp`: Timestamp for offline playback synchronization
- `incognito_mode`: Boolean indicating if the track was played in private/incognito mode

In order to read the data into a variable called `spotify`, you can use the following code:

```python
# we'll go through this in some sessions, for now just follow the steps
# save the spotify file in the same folder as this notebook
import json

with open("spotify.json", encoding="utf-8") as json_file:
    spotify = json.load(json_file)

spotify[0]
```

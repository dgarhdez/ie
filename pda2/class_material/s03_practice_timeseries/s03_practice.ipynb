Let's start by exploring our Spanish power market dataset!

```python
import pandas as pd

energy = pd.read_csv("power_market_spain_2025.csv")

energy.head()
```

## 1. Pandas for Time Series and Business Data
   - Creating and converting date columns
   - Timestamps, Periods, Timedeltas

```python
energy.dtypes
```

```python
# convert the 'datetime' column to datetime format

energy["datetime"] = pd.to_datetime(energy["datetime"])

energy["datetime"].dtype
```

```python
# create new columns for year, month, isoweek, day, hour, and date

energy["year"] = energy["datetime"].dt.year
energy["month"] = energy["datetime"].dt.month
energy["isoweek"] = energy["datetime"].dt.isocalendar().week
energy["day"] = energy["datetime"].dt.day
energy["hour"] = energy["datetime"].dt.hour
energy["date"] = energy["datetime"].dt.date

energy.head()
```

```python
# create new columns for day of the week and day and is_weekend

energy["day_of_week"] = energy["datetime"].dt.dayofweek
energy["is_weekend"] = energy["day_of_week"].isin([5, 6])  # 5 = Saturday, 6 = Sunday

energy.head()
```

```python
# create column for quarters

energy["quarter"] = energy["datetime"].dt.quarter
```

```python
energy["quarter"].unique()
```

```python
energy["month"].unique()
```

## 2. Indexing and Slicing Time Series Data
   - Date-based filtering and selection
   - Time-based data access

```python
# set 'datetime' as index minus the timezone information

energy["datetime"] = energy["datetime"].dt.tz_localize(None)

energy.set_index("datetime", inplace=True)

energy.head()
```

```python
# extract data from april 2025

energy_april_2025 = energy[(energy["year"] == 2025) & (energy["month"] == 4)]

energy_april_2025
```

   - Creating date sequences

## 5. Advanced Operations
   - Resampling and frequency conversion
   - Time shifting and lag calculations
   - Rolling windows and moving averages

```python
# plot demand in isoweek 21 hourly vs daily (resampled) vs daily (as_freq)

isoweek_21 = energy[(energy["isoweek"] == 21)]

hourly_demand_isoweek_21 = isoweek_21["demand_total"]

# plot hourly demand plotly

import plotly.express as px

fig = px.line(
    hourly_demand_isoweek_21,
    x=hourly_demand_isoweek_21.index,
    y="demand_total",
    title="Hourly Demand in Isoweek 21",
)
fig.update_xaxes(title_text="Datetime")
fig.update_yaxes(title_text="Demand (MW)")
fig.show()
```

```python
# now resample to daily frequency

daily_demand_isoweek_21 = isoweek_21["demand_total"].resample("D").mean()

# add it to the plot

fig.add_scatter(
    x=daily_demand_isoweek_21.index,
    y=daily_demand_isoweek_21,
    mode="lines",
    name="Daily Resampled Demand",
)

fig.update_layout(title="Hourly vs Daily Resampled Demand in Isoweek 21")
fig.show()
```

```python
# now use as_freq to get daily data

daily_demand_isoweek_21_as_freq = isoweek_21["demand_total"].asfreq("D")

# add it to the plot

fig.add_scatter(
    x=daily_demand_isoweek_21_as_freq.index,
    y=daily_demand_isoweek_21_as_freq,
    mode="lines",
    name="Daily As Frequency Demand",
)

fig.update_layout(title="Hourly vs Daily As Frequency Demand in Isoweek 21")
fig.show()
```

Both options are valid to extract one aggregated value per day, but they have different implications:

* `resample('D')` will aggregate the data by taking the mean of all values within each day, which is useful for continuous data like demand or prices.
* `asfreq('D')` will return the first value of each day, which is useful for discrete data or when you want to maintain the original values without aggregation.

```python
# compare the hourly spot price with the price in the previous hour using the shift method

energy["spot_price_previous_hour"] = energy["spot_price"].shift(1)

energy[["spot_price", "spot_price_previous_hour"]].head()
```

```python
# plot the two series (price and previous hour price) with plotly
# with a scatter plot in plotly with red trendline

fig = px.scatter(
    energy[["spot_price", "spot_price_previous_hour"]].dropna(),
    x="spot_price_previous_hour",
    y="spot_price",
)

# add opacity 0.3
fig.update_traces(marker=dict(opacity=0.3))

# update layout with title and axis labels
fig.update_layout(title="Spot Price vs Previous Hour Spot Price")
fig.update_xaxes(title_text="Previous Hour Spot Price (EUR/MWh)")
fig.update_yaxes(title_text="Spot Price (EUR/MWh)")
fig.show()
```

```python
# seems to be quite correlated, let's calculate the correlation coefficient

correlation = energy["spot_price"].corr(energy["spot_price_previous_hour"])

print(f"Correlation between spot price and previous hour spot price: {correlation:.2f}")
```

```python
# use rolling windows of demand to evaluate its trend over time

energy["rm24_demand"] = energy["demand_total"].rolling(window=24).mean()

# filter for isoweek 21 to plot the rolling mean
isoweek_21 = energy[(energy["isoweek"] == 21)]

# plot the demand and rolling mean with plotly in isoweek 21

fig = px.line(
    isoweek_21[["demand_total", "rm24_demand"]],
    x=isoweek_21.index,
    y=["demand_total", "rm24_demand"],
    labels={"value": "Demand (MW)", "variable": "Series"},
    title="Demand and 24-Hour Rolling Mean in Isoweek 21",
)
fig.update_xaxes(title_text="Datetime")
fig.update_yaxes(title_text="Demand (MW)")
fig.show()
```

```python
# now rolling window of 1 week to see the trend over a month of hourly data

energy["rm168_demand"] = energy["demand_total"].rolling(window=168).mean()


june_2025 = energy[(energy["year"] == 2025) & (energy["month"] == 6)]

# plot the demand and rolling mean with plotly in June 2025
fig = px.line(
    june_2025[["demand_total", "rm168_demand"]],
    x=june_2025.index,
    y=["demand_total", "rm168_demand"],
    labels={"value": "Demand (MW)", "variable": "Series"},
    title="Demand and 168-Hour Rolling Mean in June 2025",
)

fig.update_xaxes(title_text="Datetime")
fig.update_yaxes(title_text="Demand (MW)")
fig.show()
```

As we can see, rolling windows allow us to calculate moving averages, which can help smooth out fluctuations in the data and highlight longer-term trends.

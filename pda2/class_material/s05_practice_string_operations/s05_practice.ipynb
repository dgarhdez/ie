## String Operations with Pandas

We will use the pandas library with the `str` accessor to perform string operations on Series objects. This allows us to manipulate and analyze text data efficiently.

Also, we will explore regular expressions (regex) to match patterns in strings, which is a powerful tool for text processing.

For the data, we will use the Kanye Rest API, which provides a collection of Kanye West's quotes. We will extract and manipulate these quotes to demonstrate string operations.

```python
import requests

response = requests.get("https://api.kanye.rest")

response.json()
```

```python
import requests
import pandas as pd
from time import sleep

iters = range(10)
quotes = []

for i in iters:
    # get a quote from Kanye's API
    response = requests.get("https://api.kanye.rest")
    response_json = response.json()

    # add the timestamp to the response
    response_json["timestamp"] = pd.Timestamp.now()

    # append the response to the list of quotes
    quotes.append(response_json)
    print(f"Quote {i+1}: {response_json['quote']} at {response_json['timestamp']}")

    # wait for 5 seconds
    sleep(5)

quotes_df = pd.DataFrame(quotes)
```

```python
# Ana suggests thisshould be homework
# number of sentences

sentences_pattern = r"[\.\!\?]"
# breakdown of pattern
# [.!?] - matches any of the characters ., !, or ?
# + - one or more occurrences

quotes_df["sentences"] = quotes_df["quote"].str.split(sentences_pattern)
quotes_df["num_sentences"] = quotes_df["quote"].str.count(sentences_pattern) + 1
# +1 to count the last sentence that doesn't end with a punctuation mark

quotes_df.loc[3]["sentences"]
```

```python
# number of unique words and total words

quotes_df["unique_words"] = quotes_df["quote"].str.split().apply(lambda x: len(set(x)))

quotes_df
```

```python
# lexical diversity

quotes_df["lexical_diversity"] = quotes_df["unique_words"] / quotes_df["word_count"]

quotes_df
```

## Question 12: Timestamp Parsing and Time-based Analysis

Use regex to extract different components from timestamps and create a comprehensive time-based analysis of when quotes were collected.

```python
quotes_df
```

```python
# Convert timestamp column to string for regex operations
quotes_df["timestamp_str"] = quotes_df["timestamp"].astype(str)

# Extract different timestamp components using regex
quotes_df["year"] = quotes_df["timestamp_str"].str.extract(r"(\d{4})-")
quotes_df["month"] = quotes_df["timestamp_str"].str.extract(r"-(\d{2})-")
quotes_df["day"] = quotes_df["timestamp_str"].str.extract(r"-(\d{2}) ")
quotes_df["hour"] = quotes_df["timestamp_str"].str.extract(r" (\d{2}):")
quotes_df["minute"] = quotes_df["timestamp_str"].str.extract(r":(\d{2}):")
quotes_df["second"] = quotes_df["timestamp_str"].str.extract(r":(\d{2})\.")

quotes_df
```

## Pandas Vectorized String Operations

Pandas provides powerful vectorized string operations that allow you to efficiently process text data across entire Series or DataFrame columns. This is much faster and more convenient than looping through individual strings—a critical advantage when working with large datasets in business analytics and data science.

Working with text data is a common challenge in analytics and data science. Whether you're cleaning customer names, extracting information from product descriptions, or standardizing address formats, pandas' vectorized string operations provide an elegant solution.

Instead of writing loops to process each string individually, pandas allows you to apply string operations to entire Series at once, dramatically improving both performance and code readability. Let's explore how this works with practical examples.

Pandas includes features to address both this need for vectorized string operations and for correctly handling missing data via the ``str`` attribute of Pandas Series and Index objects containing strings.
So, for example, suppose we create a Pandas Series with this data:

### The `str` attribute: Your gateway to vectorized string operations

The heart of pandas string operations is the `str` attribute, available on any Series containing strings. This attribute provides access to vectorized versions of almost all Python string methods, automatically handling missing values and applying operations across the entire Series.

````python
# Example: Customer names from a July 2025 business dataset
data = ["alice johnson", "Bob Smith", None, "MARIA GARCIA", "john_doe@company.com"]
````

````python
import pandas as pd

customer_names = pd.Series(data)
customer_names
````

Now we can standardize all customer names with a single method call that automatically skips missing values—essential for real-world data cleaning:

````python
[name.lower() for name in data if name is not None]
````

````python
customer_names.str
````

````python
customer_names.str.split()
````

````python
# Compare with the traditional approach (slower and more error-prone)
processed_names = []
for name in data:
    if name is None:
        processed_names.append(None)
    else:
        processed_names.append(name.capitalize())

pd.Series(processed_names)
````

The `str` attribute automatically handles missing values and provides access to a comprehensive set of vectorized string methods. You can explore all available methods using tab completion on `customer_names.str.<TAB>` in your Jupyter environment.

## Regular Expressions: Pattern Matching Made Easy

**A regular expression (regex) is a sequence of characters that defines a search pattern for text matching and extraction.** Think of it as a powerful way to describe what you're looking for in text data, whether it's email addresses, phone numbers, product codes, or any other structured information.

Regular expressions are essential tools for data scientists and analysts working with messy, real-world text data. Python's built-in `re` module provides the interface to regular expressions. The syntax might seem complex at first, but mastering even basic patterns will dramatically improve your data cleaning and extraction capabilities.

Let's start with a practical example that demonstrates the power of regex:

#### Example: Cleaning inconsistent spacing in business data

Imagine you have a customer feedback text with inconsistent spacing (common in real-world data):

    text = "This is          a text"

Traditional string splitting fails to handle multiple spaces effectively:

````python
text = "Product quality is          excellent and delivery was fast"
````

````python
# Traditional approach creates empty strings from multiple spaces
text.split(" ")
````

Regular expressions provide an elegant solution by finding patterns, and then either splitting, replacing, extracting, or validating text based on those patterns. In this case, we can use a regex pattern to match one or more spaces and replace them with a single space:

````python
# Regex pattern: \s+ means "one or more whitespaces"
# Raw string (r'') prevents Python from interpreting backslashes
pattern = r"\s+"
print(f"Pattern: {pattern}")
print("\\s = any whitespace (space, tab, newline)")
print("+ = one or more of the preceding character")
````

````python
text
````

````python
import re

# Compile the regex pattern for reuse (more efficient)
regex = re.compile(r"\s+")

# Split text on any sequence of whitespace characters
cleaned_words = regex.split(text)
cleaned_words
````

````python
text
````

Perfect! The regex pattern `\s+` successfully handles any amount of whitespace. Here's what happened:

- `\s` matches any whitespace character (space, tab, newline, etc.)
- `+` means "one or more" of the preceding element
- Together, `\s+` matches any sequence of one or more whitespace characters

This is much more powerful than simple string splitting because it handles inconsistent spacing automatically.

#### Another practical example: Finding specific patterns

````python
description = """
Our new laptop model ABC is fast and reliable,
model ABC features great performance
"""
````

````python
print(description)
````

````python
# Find pattern: space + exactly 3 word characters + space
# This will match " ABC " (including surrounding spaces)
pattern = re.compile(r"\s\w{3}\s")
result = pattern.split(description)
result
````

The pattern `\s\w{3}\s` breaks down as:
- `\s` = whitespace character
- `\w{3}` = exactly three alphanumeric characters (word characters)
- `\s` = whitespace character

The regex found both instances of " ABC " in our product description and split the text at those locations.

In the `re` package, a regular expresion must be compiled first with `compile`, turning it into a `Pattern` object. The type of this object is `re.Pattern`. You can then use this pattern to search, match, or manipulate text data efficiently.

````python
type(regex)
````

There is a lot of string operations we can do with a regular expression. Check the full list with `regex.` tab completion. The most common
 - `findall` - Return a list of all non-overlapping matches of pattern in string.
 - `search` - Scan through string looking for a match, and return a corresponding match object instance

## Advanced Regular Expression Concepts

### A business-critical example: Email validation

We can assume that emails are made up of:

* Some alphanumeric characters (letters, numbers, underscores)
* An `@` symbol
* More alphanumeric characters (domain name)
* A dot (`.`)
* A top-level domain (TLD) consisting of exactly three lowercase letters

````python
import re

email_pattern = re.compile(r"\w+@\w+\.[a-z]{3}")  # this is the pattern of an email
````

The pattern can be broken down as follows:

* `\w+` = one or more characters (alphanumeric + underscore)
* `@` = literal `@` symbol
* `\w+` = one or more characters (domain)
* `\.` = literal dot (escaped)
* `[a-z]{3}` = exactly 3 lowercase letters (TLD)

Let's extract email addresses from business communications or customer data:

````python
# Extract emails from business communication (July 2025)
business_text = (
    "Contact our sales team at sales1234@company.com or reach support at help@support.org"
)
email_pattern.findall(business_text)
````

But what about emails like `daniel.garcia@domain.co`? We need to improve our pattern to handle dots in the username and allow for TLDs with 2 or 3 characters:

````python
improved_email = re.compile(r"[\w.]+@[\w.]+\.[a-z]{2,3}")
improved_email.findall("sarah.johnson@company.ie.com")
````

The new pattern can be broken down as follows:

* `[\w.]+` = groups of one or more alphanumeric characters or dots
* `@` = literal `@` symbol
* `[\w.]+` = groups of one or more alphanumeric characters or dots (domain)
* `\.` = literal dot (escaped)
* `[a-z]{2,3}` = exactly 2 or 3 lowercase letters (TLD)

````python
email = re.compile(r"[\w.]+@[\w.]+\.[a-z]{2,3}")
````

````python
# Define the improved email pattern first
improved_email = re.compile(r"[\w.]+@[\w.]+\.[a-z]{2,3}")

# Test with various business email formats
test_emails = ["sarah.johnson@company.com", "mike@sales.io", "team@support.org"]
for email_address in test_emails:
    print(f"{email_address}: {improved_email.findall(email_address)}")
````

### Metacharacters: The power of regex

Certain characters have special meanings in regular expressions. These **metacharacters** are what make regex so powerful for pattern matching:

```
. ^ $ * + ? { } [ ] \ | ( )
```

Here's a practical guide to the most important metacharacters for business applications:

| Character | Description | Business Example |
|-----------|-------------|------------------|
| ``.`` | Matches any character except newline | ``prod.ct`` matches "product", "prod3ct" |
| ``^`` | Start of string | ``^Order`` matches "Order" only at the beginning |
| ``$`` | End of string | ``\.com$`` matches ".com" only at the end |
| ``*`` | Zero or more repetitions | ``colou*r`` matches "color" and "colour" |
| ``+`` | One or more repetitions | ``\d+`` matches any number (one or more digits) |
| ``?`` | Zero or one repetition | ``colou?r`` matches "color" or "colour" |
| ``{m}`` | Exactly m repetitions | ``\d{4}`` matches exactly 4 digits (like years) |
| ``{m,n}`` | Between m and n repetitions? | ``\d{3,5}`` matches 3, 4, or 5 digits |
| ``[ ]`` | Character set | ``[Tt]otal`` matches "Total" or "total" |
| ``\`` | Escape special characters | ``\$`` matches literal "$" symbol |
| ``\|`` | OR operation | ``cat\|dog`` matches "cat" or "dog" |
| ``( )`` | Groups for extraction | ``(\d+)-(\d+)`` captures two numbers separated by dash |

To match these special characters literally, escape them with a backslash:

````python
regex = re.compile(r"\$")
regex.findall("Product costs $29.99, shipping is $5.00")
````

### Interlude: raw strings

In Python, regular expressions are often written as raw strings to avoid confusion with escape sequences. This is done by prefixing the string with `r`, like `r'\d{3}'`. This tells Python to treat backslashes as literal characters, which is essential for regex patterns.

A raw string is prefixed with an `r` and treats backslashes as literal characters, making it easier to write regex patterns without excessive escaping.

````python
print("a\tb")
````

````python
print(r"a\tb")
````

For this reason, whenever you use backslashes in a regular expression, it is good practice to use a raw string.

#### Putting together several characters to build the search pattern

````python
regex = re.compile(r"[\w]+\s[\w]+")
regex.findall("the fox is 9 years old")
````

In this case, the '\w\s\w' pattern matches an alphanumerical character (`\w`), followed by a whitespace (`\s`), followed by another alphanumerical character (`\w`).

The following table lists a few of these characters that are commonly useful:

| Character | Description                 | Character | Description                     |
|-----------|-----------------------------|-----------|---------------------------------|
| ``"\d"``  | Match any digit             | ``"\D"``  | Match any non-digit             |
| ``"\s"``  | Match any whitespace        | ``"\S"``  | Match any non-whitespace        |
| ``"\w"``  | Match any alphanumeric char | ``"\W"``  | Match any non-alphanumeric char |

This is *not* a comprehensive list or description; for more details, see Python's [regular expression syntax documentation](https://docs.python.org/3/library/re.html#re-syntax).

````python
regex = re.compile(r"\W\S")
regex.findall("!a2%e4")
````

#### Square brackets match custom character groups

If the built-in character groups aren't specific enough for you, you can use square brackets to specify any set of characters you're interested in.
For example, the following will match any lower-case vowel:

````python
list_of_vowels = "aeiou"
word = "consequential"

# Method 1: Using list comprehension to find vowels
vowels_found = [letter for letter in word if letter in list_of_vowels]
print(f"Vowels found in '{word}': {vowels_found}")
````

````python
# Method 2: Using regex (more relevant to this section)
vowel_regex = re.compile("[aeiou]")
vowels_regex = vowel_regex.findall(word)
print(f"Vowels found using regex: {vowels_regex}")
````

````python
# to get the characters within the vowels
regex = re.compile("[aeiou]")
regex.split("consequential")
````

Similarly, you can use a dash to specify a range: for example, ``"[a-z]"`` will match any lower-case letter, and ``"[1-3]"`` will match any of ``"1"``, ``"2"``, or ``"3"``.
For instance, you may need to extract from a document specific numerical codes that consist of a capital letter followed by a digit. You could do this as follows:

````python
# find combinations of 1 upper case letter and 1 number
regex = re.compile("[A-Z][0-9]")

regex.findall("104A3879, G2, H6")
````

#### Wildcards match repeated characters

If you would like to match a string with, say, three alphanumeric characters in a row, it is possible to write, for example, ``"\w\w\w"``.
Because this is such a common need, there is a specific syntax to match repetitions – curly braces with a number:

````python
regex = re.compile(r"\w{3,5}")
regex.findall("The quick brownnnnnnn foxx")
````

There are also markers available to match any number of repetitions – for example, the ``"+"`` character will match *one or more* repetitions of what precedes it:

````python
regex = re.compile(r"\w+")
regex.findall("The quick brown fox")
````

The following is a table of the repetition markers available for use in regular expressions:

| Character | Description | Example |
|-----------|-------------|---------|
| ``?`` | Match zero or one repetitions of preceding  | ``"ab?"`` matches ``"a"`` or ``"ab"`` |
| ``*`` | Match zero or more repetitions of preceding | ``"ab*"`` matches ``"a"``, ``"ab"``, ``"abb"``, ``"abbb"``... |
| ``+`` | Match one or more repetitions of preceding  | ``"ab+"`` matches ``"ab"``, ``"abb"``, ``"abbb"``... but not ``"a"`` |
| ``{n}`` | Match ``n`` repetitions of preeeding | ``"ab{2}"`` matches ``"abb"`` |
| ``{m,n}`` | Match between ``m`` and ``n`` repetitions of preceding | ``"ab{2,3}"`` matches ``"abb"`` or ``"abbb"`` |

````python
regex = re.compile(r"\w{3,4}")
regex.findall("The quick brown fox")
````

#### Parentheses indicate *groups* to extract

For compound regular expressions like our email matcher, we often want to extract their components rather than the full match. This can be done using parentheses to *group* the results.

Whatever matches the pattern inside the parentheses will be captured and can be extracted separately.

````python
email3 = re.compile(r"[\w.]+@(\w+)\.[a-z]{3}")

# dani@dani.com
````

The pattern `'([\w.]+)@(\w+)\.([a-z]{3})'` has three groups:
* The first group is any alphanumeric character or a dot, repeated one or more times: `([\w.]+)`
* The second group is any alphanumeric character, repeated one or more times: `(\w+)`
* The third group is any lowercase letter, repeated three times: `([a-z]{3})`

Basically, we are finding strings that look like emails, and then extracting the username, domain name and top-level domain.

````python
text = "To email Guido, try guido@python.org or the older address guido@google.com."
email3.findall(text)
````

As we see, this grouping actually extracts a list of the sub-components of the email address.

We can go a bit further and *name* the extracted components using the ``"(?P<name> )"`` syntax, in which case the groups can be extracted as a Python dictionary:

````python
email4 = re.compile(r"(?P<key1>\w+)@(?P<key2>\w+)\.(?P<key3>[a-z]{3})")
match = email4.match("guido@python.org")
match.groupdict()
````

## Pandas String Methods with Regular Expressions

Pandas seamlessly integrates regular expressions with its string methods, providing powerful tools for data cleaning and extraction at scale. Here are the essential regex-enabled methods for business analytics:

| Method | Description | Use Case |
|--------|-------------|----------|
| ``match()`` | Check if string starts with pattern | Validate format compliance |
| ``extract()`` | Extract matched groups as DataFrame columns | Split structured data |
| ``findall()`` | Find all matches in each string | Extract multiple items |
| ``replace()`` | Replace pattern matches | Clean and standardize data |
| ``contains()`` | Check if pattern exists anywhere | Filter and categorize |
| ``count()`` | Count pattern occurrences | Frequency analysis |
| ``split()`` | Split on regex pattern | Parse delimited data |

Let's explore practical business applications:

With these, you can do a wide range of interesting operations.
For example, we can extract the first name from each by asking for a contiguous group of characters at the beginning of each element:

````python
customers = pd.Series(
    [
        "Sarah Williams - sarah.williams@company.com",
        "Michael Chen - m.chen@business.org",
        "Emma Rodriguez - emma.r@startup.io",
        "David Kim - david.kim@corporation.com",
    ]
)
````

````python
customers
````

````python
customers.str.extract(r"([^-]+)")[0].values[0].rstrip() # extracts characters until the first dash
````

````python
customers.str.extract(
    r"([A-Za-z]+)"
)  # 1 or more letters upper or lower case
````

````python
# Find all email addresses in customer data
customers.str.findall(r"[\w.]+@[\w.]+\.[a-z]{2,3}")
````

#### Advanced pattern matching: Business filtering

Let's create a more sophisticated example for filtering business data:

````python
companies = pd.Series(["Amazon Tech", "Google Inc", "Microsoft Corp", "Oracle Systems"])

companies
````

````python
# Check if company names start with a vowel and end with a consonant
pattern = r"^[AEIOUaeiou].*[bcdfghjklmnpqrstvwxyzBCDFGHJKLMNPQRSTVWXYZ]$"
companies.str.match(pattern)
````

````python
# Find companies that contain 'Tech', 'Inc', 'Corp', or 'Systems'
companies.str.contains(r"(Tech|Inc|Corp|Systems)")
````

## Getting Started

### Step 1: Get Your API Key

1. Visit https://openweathermap.org/api
2. Sign up for a free account
3. Navigate to your API keys section
4. Copy your API key

**Note:** It may take a few minutes for your API key to be activated after signing up.

### Step 2: Set Up Your .env File

1. Create a file named `.env` in the same directory as this notebook
2. Add your API key: `OPENWEATHER_API_KEY=your_key_here`
3. Save the file

**IMPORTANT:** Never commit your `.env` file to version control! Add it to `.gitignore` to keep your API key secure.

````python
import requests
import pandas as pd
from dotenv import load_dotenv
import os
from datetime import datetime
import time

# Load environment variables from .env file
load_dotenv()

# Get the API key
API_KEY = os.getenv('OPENWEATHER_API_KEY')

# Verify the key is loaded
if API_KEY:
    print("✓ API key loaded successfully!")
    print(f"Key preview: {API_KEY[:8]}...")
else:
    print("✗ ERROR: API key not found. Please check your .env file.")

# Base URL for OpenWeatherMap API
BASE_URL = "https://api.openweathermap.org/data/2.5"
````

## Testing the API Connection

Let's verify our API connection works by fetching current weather for a city.

````python
def get_current_weather(city):
    """
    Fetch current weather for a city.
    
    Args:
        city (str): City name
    
    Returns:
        dict or None: Weather data if successful
    """
    endpoint = f"{BASE_URL}/weather"
    params = {
        'q': city,
        'appid': API_KEY,
        'units': 'metric'  # Use Celsius
    }
    
    response = requests.get(endpoint, params=params)
    
    if response.status_code == 200:
        return response.json()
    else:
        print(f"Error {response.status_code}: {response.reason}")
        return None

# Test with Madrid
madrid_weather = get_current_weather('Madrid')
if madrid_weather:
    print(f"Temperature in Madrid: {madrid_weather['main']['temp']}°C")
    print(f"Weather: {madrid_weather['weather'][0]['description']}")
    print(f"Humidity: {madrid_weather['main']['humidity']}%")
    print(f"Wind Speed: {madrid_weather['wind']['speed']} m/s")
````

# Introduction to APIs

## What is an API?

An API, or Application Programming Interface, is a set of rules and protocols that allows different software applications to communicate with each other. Think of an API as a bridge that enables two applications to exchange information and perform tasks without needing to understand the internal workings of each other.

For example, when you use a weather app on your phone, the app communicates with a weather service's API to fetch the latest weather data. The API defines how the app can request the data and how the weather service will respond.

## HTTP Protocol

The Hypertext Transfer Protocol (HTTP) is the foundation of data communication on the web. It is a protocol used for transmitting hypertext requests and information between clients (such as web browsers) and servers. HTTP defines a set of request methods that indicate the desired action to be performed on a given resource.

### Common HTTP Request Types

- **GET**: Requests data from a specified resource. For example, fetching a webpage or retrieving data from an API.
- **POST**: Submits data to be processed to a specified resource. For example, submitting a form or uploading a file.
- **PUT**: Updates a specified resource with new data. For example, updating a user's profile information.
- **DELETE**: Deletes a specified resource. For example, removing a user account.

### Common HTTP Response Codes

When you make an HTTP request to a web API, the server responds with an HTTP response. The response typically includes a status code, headers, and a body. HTTP status codes are three-digit numbers that indicate the result of the HTTP request. Here are some of the most relevant status codes:

- **200 OK**: The request was successful, and the server returned the requested data.
- **201 Created**: The request was successful, and a new resource was created.
- **204 No Content**: The request was successful, but there is no content to return.
- **400 Bad Request**: The server could not understand the request due to invalid syntax.
- **401 Unauthorized**: The client must authenticate itself to get the requested response.
- **403 Forbidden**: The client does not have access rights to the content.
- **404 Not Found**: The server cannot find the requested resource.
- **500 Internal Server Error**: The server encountered an unexpected condition that prevented it from fulfilling the request.

## Web APIs

Web APIs allow different software applications to communicate over the internet. They enable developers to access web services and integrate various functionalities into their applications by making HTTP requests to specific endpoints.

### Connecting to Web APIs

To connect to a web API, you typically need to:

1. **Find the API Endpoint**: The endpoint is the URL where the API can be accessed. For example, `https://api.example.com/v1/weather`.
2. **Make an HTTP Request**: Use an HTTP client (such as a web browser, Postman, or a programming language library) to send a request to the API endpoint. Specify the HTTP method (GET, POST, etc.) and any required parameters or headers.
    - In Python we can use the `requests` library to make HTTP requests. If not installed, you can install it using `pip install requests` in your environment.
3. **Handle the Response**: The API will respond with data, usually in **JSON format**. Process the response data as needed in your application.

Here is a simple example of making a GET request to a web API using Python:

```python
import requests

response = requests.get('https://api.example.com/v1/weather?city=London')
if response.status_code == 200:
    data = response.json()
    print(data)
else:
    print('Failed to retrieve data:', response.status_code)
```

In the URL we have these parts:

- `https://api.example.com/v1` is the base URL
- `/weather` is the endpoint
- `city=London` is the query parameter

In this example, we use the `requests` library to send a GET request to the weather API endpoint, specifying the city as a query parameter. We then check the response status code and print the data if the request was successful.


### Authentication in Web APIs

Authentication is a crucial aspect of Web APIs as it ensures that only authorized users can access and interact with the resources provided by the API. There are several methods to implement authentication in Web APIs, including:

1. **API Keys**: A unique key is provided to each user, which must be included in the request headers or query parameters. This key is used to identify and authenticate the user.

2. **Basic Authentication**: The user's credentials (username and password) are encoded in Base64 and included in the request headers. This method is simple but less secure as the credentials are exposed in each request.

3. **OAuth**: A more secure and flexible method that allows users to grant third-party applications limited access to their resources without sharing their credentials. OAuth involves the use of access tokens that are obtained through an authorization process.

4. **Bearer Token**: Similar to API keys, bearer tokens are included in the request headers. These tokens are usually generated by an authentication server and have a limited lifespan.

5. **JWT (JSON Web Tokens)**: A compact and self-contained way to transmit information between parties as a JSON object. JWTs are signed using a secret or a public/private key pair, ensuring the integrity and authenticity of the token.

Implementing proper authentication mechanisms in your Web API is essential to protect sensitive data and ensure that only authorized users can perform specific actions.

## Example: Connecting to the PokeAPI

The PokeAPI (https://pokeapi.co/) is a free, open-source Web API that provides comprehensive data about Pokemon, including information about Pokemon species, abilities, moves, types, and more. It's a great example of a well-documented public API that doesn't require authentication.

### Accessing the PokeAPI

To connect to the PokeAPI:

1. **Base URL**: The main endpoint is `https://pokeapi.co/api/v2/`

2. **No Authentication Required**: Unlike many APIs, PokeAPI doesn't require an API key or any form of authentication

3. **Make API Requests**: Use an HTTP client (such as `requests` in Python) to send requests to the PokeAPI endpoints.

4. **Handle the Responses**: The API will respond with data in JSON format. Process the response data as needed in your application.

We'll use the Python requests library to interact with this API and demonstrate how to retrieve Pokemon data.


````python
import requests

# Define the base URL for the PokeAPI
base_url = "https://pokeapi.co/api/v2"

# Define the endpoint to retrieve the list of Pokemon
endpoint_url = f"{base_url}/pokemon/1"   # --> This is to get the first pokemon

# Make a GET request to retrieve the list of Pokemon
response = requests.get(endpoint_url)

# Check if the request was successful
if response.status_code == 200:
    # Parse the JSON response
    pokemon_data = response.json()
else:
    print(f"Failed to retrieve data. Status code: {response.status_code}")
````

````python
pprint.pprint(pokemon_data)
````

***Exercise 1: Use the PokeAPI to access the first 20 pokemon. Then build a DataFrame with 4 columns: name, height, weight and type (pick the first)***  

````python
import requests
import pandas as pd

# Initialize lists to store the data
names = []
heights = []
weights = []
types = []

base_url = "https://pokeapi.co/api/v2"

# Loop through first 20 pokemon
for i in range(1, 21):
    # Get data for each pokemon
    endpoint_url = f"{base_url}/pokemon/{i}"
    response = requests.get(endpoint_url)
    
    if response.status_code == 200:
        pokemon_data = response.json()
        
        # Extract the required information
        names.append(pokemon_data['name'])
        heights.append(pokemon_data['height'])
        weights.append(pokemon_data['weight'])
        types.append(pokemon_data['types'][0]['type']['name'])  # Get first type
    else:
        print(f"Failed to retrieve data for pokemon {i}. Status code: {response.status_code}")

# Create DataFrame
pokemon_df = pd.DataFrame({
    'name': names,
    'height': heights, 
    'weight': weights,
    'type': types
})

# Display the DataFrame
pokemon_df
````

### Good practice for Access Keys

When working with APIs, it's important to keep your secret access keys secure. One way to do this is by using environment variables. Environment variables allow you to store sensitive information outside of your codebase, reducing the risk of accidentally exposing your keys.

Here's how you can use environment variables in Python:

1. **Set up environment variables**: You can set environment variables in your operating system or in a `.env` file. For example, create a `.env` file in your project directory and add your API key:

```
TMDB_API_KEY=your_auth_token_here
```

2. **Load environment variables in your Python code**: Use the `python-dotenv` package to load the environment variables from the `.env` file. First, install the package if you haven't already:

```
pip install python-dotenv
```

Then, load the environment variables in your code

By using environment variables, you can keep your API keys secure and avoid hardcoding them in your source code.



````python
from dotenv import load_dotenv
import os

# Load environment variables from .env file
load_dotenv("variables.env")

# Get the API key from the environment variable
auth_token = os.getenv('TMDB_API_KEY')

# Define the API endpoint
api_endpoint = 'https://api.themoviedb.org/3/movie/550'

# Set up the headers with the Bearer token for authentication
headers = {
    'Authorization': f'Bearer {auth_token}'
}

# Make the GET request to the TMDB API
response = requests.get(api_endpoint, headers=headers)

# Check the response status code and handle the response
if response.status_code == 200:
    data = response.json()
    pprint.pprint(data)
else:
    print('Failed to retrieve data:', response.status_code)
````

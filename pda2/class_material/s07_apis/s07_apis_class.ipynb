## What is an API?

An **API (Application Programming Interface)** is a set of rules and protocols that allows different software applications to communicate with each other. Think of an API as a bridge that enables two applications to exchange information and perform tasks without needing to understand the internal workings of each other.

### REST APIs

REST (Representational State Transfer) is an architectural style for designing networked applications. REST APIs use HTTP requests to access and manipulate data.

### HTTP Methods

- **GET**: Retrieve data from a server (we'll focus on this)
- **POST**: Send data to create a new resource
- **PUT**: Update an existing resource
- **DELETE**: Remove a resource

### HTTP Status Codes

| Code | Meaning | Description |
|------|---------|-------------|
| 200 | OK | Request successful |
| 201 | Created | Resource successfully created |
| 400 | Bad Request | Invalid request syntax |
| 401 | Unauthorized | Authentication required |
| 404 | Not Found | Resource doesn't exist |
| 500 | Internal Server Error | Server encountered an error |

- JSON objects → Python dictionaries `{}`

API responses often contain nested structures. Let's extract specific data from the Pokemon response.

````python
# Get Pokemon abilities
pokemon_name = "bulbasaur"
response = requests.get(f"{base_url}/pokemon/{pokemon_name}")
data = response.json()

# Extract abilities (nested in a list of dictionaries)
abilities = [ability["ability"]["name"] for ability in data["abilities"]]
print(f"{pokemon_name.title()} abilities: {abilities}")

# Extract types
types = [type_data["type"]["name"] for type_data in data["types"]]
print(f"{pokemon_name.title()} types: {types}")
````

## Python Exception Handling: Try-Except Blocks

Before we dive into API error handling, let's understand how Python handles errors using **try-except blocks**.

### What are Exceptions?

An **exception** is an error that occurs during program execution. When Python encounters an error, it "raises" an exception and stops the program unless you handle it.

Common exceptions:
- `TypeError`: Wrong data type
- `ValueError`: Correct type but inappropriate value
- `KeyError`: Dictionary key doesn't exist
- `IndexError`: List index out of range
- `ZeroDivisionError`: Division by zero
- `FileNotFoundError`: File doesn't exist

### Why Use Try-Except?

Without exception handling, a single error stops your entire program. With try-except blocks, you can:
- **Gracefully handle errors** instead of crashing
- **Provide user-friendly messages** instead of technical error traces
- **Continue execution** even when errors occur
- **Log errors** for debugging while keeping the program running

### Basic Try-Except Syntax

```python
try:
    # Code that might raise an exception
    result = 10 / 0
except ZeroDivisionError:
    # Code to handle the specific error
    print("Cannot divide by zero!")
```

### Example: Handling Multiple Exceptions

```python
try:
    user_input = input("Enter a number: ")
    number = int(user_input)
    result = 10 / number
    print(f"Result: {result}")
    
except ValueError:
    print("That's not a valid number!")
    
except ZeroDivisionError:
    print("Cannot divide by zero!")
    
except Exception as e:
    # Catch any other unexpected errors
    print(f"An unexpected error occurred: {e}")
```

### Try-Except-Else-Finally

Python provides additional clauses for more control:

```python
try:
    # Code that might raise an exception
    file = open("data.txt", "r")
    data = file.read()
    
except FileNotFoundError:
    print("File not found!")
    
else:
    # Executes if NO exception occurred
    print("File read successfully!")
    
finally:
    # ALWAYS executes, regardless of exceptions
    # Useful for cleanup (closing files, connections, etc.)
    if 'file' in locals():
        file.close()
        print("File closed")
```

### Best Practices

1. **Be specific**: Catch specific exceptions rather than using a broad `except Exception`
2. **Don't hide errors**: Log or report what went wrong
3. **Keep try blocks small**: Only wrap code that might raise exceptions
4. **Clean up resources**: Use `finally` or context managers (`with` statement)
5. **Re-raise when appropriate**: Sometimes you need to handle an error partially and let it propagate

### Why This Matters for APIs

When working with APIs, many things can go wrong that are outside your control.

Using try-except blocks ensures your program handles these issues gracefully instead of crashing!

Let's now see how to apply this to API requests.

## Error Handling in API Calls

When working with APIs, many things can go wrong:
- **Network failures**: Internet connection issues
- **Invalid endpoints**: Wrong URL or Pokemon name
- **Rate limiting**: Too many requests in a short time
- **Malformed data**: Unexpected response structure

Always implement proper error handling!

````python
def get_pokemon_safe(pokemon_name):
    """
    Safely fetch Pokemon data with comprehensive error handling.

    Args:
        pokemon_name (str): Name of the Pokemon

    Returns:
        dict or None: Pokemon data if successful, None otherwise
    """
    try:
        url = f"{base_url}/pokemon/{pokemon_name.lower()}"
        response = requests.get(url, timeout=10)

        # Check if request was successful
        if response.status_code == 200:
            return response.json()
        elif response.status_code == 404:
            print(f"Pokemon '{pokemon_name}' not found.")
            return None
        else:
            print(f"Error {response.status_code}: {response.reason}")
            return None

    except requests.exceptions.Timeout:
        print("Request timed out. Please try again.")
        return None
    except requests.exceptions.ConnectionError:
        print("Connection error. Check your internet connection.")
        return None
    except Exception as e:
        print(f"Unexpected error: {e}")
        return None


# Test with valid and invalid names
print("Valid Pokemon:")
data = get_pokemon_safe("pikachu")
if data:
    print(f"Successfully fetched {data['name']}")

print("\nInvalid Pokemon:")
get_pokemon_safe("invalidpokemon")
````

## Building DataFrames from API Data

One of the most powerful uses of APIs is collecting data for analysis. Let's fetch multiple Pokemon and create a DataFrame.

### Why Convert to DataFrame?

- **Structured format**: Easy to analyze and visualize
- **Pandas operations**: Filter, group, aggregate, etc.
- **Export options**: Save to CSV, Excel, databases

````python
# Fetch first 20 Pokemon
pokemon_list = []

for i in range(1, 21):
    response = requests.get(f"{base_url}/pokemon/{i}")

    if response.status_code == 200:
        data = response.json()

        pokemon_list.append(
            {
                "id": data["id"],
                "name": data["name"],
                "height": data["height"],
                "weight": data["weight"],
                "base_experience": data["base_experience"],
                "type_primary": data["types"][0]["type"]["name"],
                "type_secondary": (
                    data["types"][1]["type"]["name"] if len(data["types"]) > 1 else None
                ),
            }
        )
    else:
        print(f"Failed to fetch Pokemon {i}")

# Create DataFrame
pokemon_df = pd.DataFrame(pokemon_list)
pokemon_df
````

````python
# Basic DataFrame analysis
print("Descriptive Statistics:")
print(pokemon_df.describe())

print("\nCount by Primary Type:")
print(pokemon_df["type_primary"].value_counts())

print("\nHeight-Weight Correlation:")
print(f"{pokemon_df['height'].corr(pokemon_df['weight']):.3f}")
````

## Working with Different Endpoints

PokeAPI provides multiple endpoints for different types of data:

- `/pokemon/{id or name}` - Pokemon details
- `/type/{id or name}` - Type information
- `/ability/{id or name}` - Ability details
- `/pokemon-species/{id or name}` - Species data (descriptions, evolution chains)

Let's explore some of these!

````python
# Get all Pokemon of type "fire"
response = requests.get(f"{base_url}/type/fire")
data = response.json()

# Extract Pokemon names with this type (first 10)
fire_pokemon = [pokemon["pokemon"]["name"] for pokemon in data["pokemon"][:10]]
print(f"First 10 Fire-type Pokemon: {fire_pokemon}")
````

````python
def get_pokemon_with_species_info(pokemon_name):
    """
    Fetch Pokemon data and species information by combining multiple endpoints.
    """
    # Get basic Pokemon data
    pokemon_response = requests.get(f"{base_url}/pokemon/{pokemon_name}")
    pokemon_data = pokemon_response.json()

    # Get species data (contains description)
    species_url = pokemon_data["species"]["url"]
    species_response = requests.get(species_url)
    species_data = species_response.json()

    # Extract English flavor text (description)
    english_flavor = next(
        (
            entry["flavor_text"]
            for entry in species_data["flavor_text_entries"]
            if entry["language"]["name"] == "en"
        ),
        "No description available",
    )

    return {
        "name": pokemon_data["name"],
        "height": pokemon_data["height"],
        "weight": pokemon_data["weight"],
        "generation": species_data["generation"]["name"],
        "description": english_flavor.replace("\n", " ").replace("\f", " "),
    }


# Test the function
pprint.pprint(get_pokemon_with_species_info("pikachu"))
````

## Query Parameters

Query parameters allow you to modify API requests by adding information to the URL.

**Format:** `https://api.example.com/endpoint?param1=value1&param2=value2`

**Common uses:**
- Pagination (limit, offset)
- Filtering (filter by certain criteria)
- Sorting (order results)

In the `requests` library, use the `params` argument to pass query parameters.

````python
# Fetch Pokemon list with pagination
response = requests.get(f"{base_url}/pokemon", params={"limit": 10, "offset": 20})

data = response.json()

# Show results
pokemon_names = [pokemon["name"] for pokemon in data["results"]]
print(f"Pokemon 21-30: {pokemon_names}")
print(f"Total count: {data['count']}")
print(f"Next page URL: {data['next']}")
````

In the previous call, the params dictionary was used to specify pagination options:
```python
params = {'limit': 10, 'offset': 20}
```

This means "give me 10 results, starting from the 21st item".

### Implementing Simple Caching

Caching stores previously fetched data so you don't need to make the same API call multiple times.

````python
import time

# Simple cache dictionary
pokemon_cache = {}


def get_pokemon_cached(pokemon_name):
    """
    Fetch Pokemon data with simple caching.
    """
    if pokemon_name in pokemon_cache:
        print(f"Returning cached data for {pokemon_name}")
        return pokemon_cache[pokemon_name]

    print(f"Fetching data for {pokemon_name} from API")
    response = requests.get(f"{base_url}/pokemon/{pokemon_name}")

    if response.status_code == 200:
        data = response.json()
        pokemon_cache[pokemon_name] = data
        return data
    return None


# Test caching
print("First call:")
start = time.time()
get_pokemon_cached("pikachu")
print(f"Time: {time.time() - start:.3f}s")

print("\nSecond call (cached):")
start = time.time()
get_pokemon_cached("pikachu")
print(f"Time: {time.time() - start:.6f}s")
````

## Filter Data Above Threshold

Create a function `filter_above_threshold(data, threshold)` that takes a list of numbers and a threshold, returning a list of numbers above the threshold.

Create a test function `test_filter_above_threshold()` that verifies the correctness of `filter_above_threshold(data, threshold)` using assertions with known inputs and expected outputs.

You should check for:
- Normal filtering case
- Threshold higher than all elements (empty result)
- Threshold lower than all elements (all elements returned)
- Empty input list
- List with duplicate values
- Negative numbers and thresholds
- Floating-point numbers and thresholds

```python
# [1, 2, 3], threshold = 1
# [2, 3]


def filter_above_threshold(data, threshold):

    if not data:
        raise ValueError("The list cannot be empty")

    result = []
    for number in data:
        if number > threshold:
            result.append(number)
    return result


filter_above_threshold([1, 2, 3], 1)
```

## Compute Standard Deviation

```python
import numpy as np

np.std([1, 2, 3, 4])
```

```python
def compute_std(data):
    n = len(data)
    mean_data = calculate_mean(data)
    delta = 0
    for element in data:
        delta += (element - mean_data) ** 2
    if n < 30:
        std = (delta / (n - 1)) ** 0.5
    else:
        std = (delta / n) ** 0.5
    return std


compute_std([1, 2, 3, 4])
```

## Merge Two Lists into Dictionary

Create a function `merge_to_dict(keys, values)` that takes two lists and returns a dict.

Create a test function `test_merge_to_dict()` that verifies the correctness of `merge_to_dict(keys, values)` using assertions with known inputs and expected outputs.

You should check for:
- Equal length lists
- Keys list longer than values (should raise exception or handle gracefully)
- Values list longer than keys (should raise exception or handle gracefully)
- Empty lists
- Lists with duplicate keys
- Different data types in lists

```python
def merge_to_dict(keys, values):

    if len(keys) != len(values):
        raise ValueError("The lists must have equal lengths")

    return dict(zip(keys, values))


merge_to_dict([1, 2, 3], [2, 3, 4])
```
